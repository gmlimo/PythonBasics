## Step 2: Connect to Your API

Create a function to make API requests with proper error handling.


```python
# Create your API connection function
# Include error handling with try/except
# Test it with a simple request

# Your code here

# Fetch the webpage content
url = 'https://arxiv.org/list/math-ph/2026-01'
data = requests.get(url)
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"}

try:
    response = requests.get(url, headers = headers, timeout = 15)
    response.raise_for_status()
    
except requests.exceptions.RequestsException as e:
    print('Error fecthing data: {e}')
    exit()
    
response.status_code
```

```python
# Parse HTML content using BeautifulSoup
soup = BeautifulSoup(response.text, "html.parser")

# arXiv groups list items in <dt> (identifiers/links) and <dd> (title/authors)
dt_tags = soup.find_all("dt")
dd_tags = soup.find_all("dd")

records = []
```

## Step 3: Collect Your Data

Write code to collect at least 50 records from your API. Include rate limiting (time delays) to be respectful to the API.


```python
# Write your data collection function
# Collect at least 50 records
# Add time.sleep() between requests
# Store data in a list

# Your code here

# Extract and match invidual metadata elements
for dt, dd in zip(dt_tags, dd_tags):
    # Extract the arXiv ID (e.g., "arXiv:20601.00113")
    arxiv_id_tag = dt.find("a", title = "Abstract")
    arxiv_id = arxiv_id_tag.text.strip() if arxiv_id_tag else np.nan
    
    # Extract the Title
    title_div = dd.find("div", class_ = "list-title")
    title = (title_div.text.replace("Title:", "").strip() if title_div else np.nan)
    
    # Extract the Authors
    authors_div = dd.find("div", class_ = "list-authors")
    authors = (authors_div.text.replace("Authors:", "").strip() if authors_div else np.nan)
    
    # Extract the Subjects
    subjects_div = dd.find("div", class_ = "list-subjects")
    subjects = (subjects_div.text.replace("Subjects:", "").strip() if subjects_div else np.nan)
    
    records.append(
            {
               "arxiv_id": arxiv_id,
               "title": title,
               "authors": authors,
               "subjects": subjects,
            }
                  )
```

## Step 4: Process Your Data with Pandas

Convert your collected data to a DataFrame and clean it up.


```python
# Convert your data to a pandas DataFrame
# Check for missing values
# Clean any messy text data
# Check the shape and basic info

# Your code here

df = pd.DataFrame(records)

# Use numpy to handle any missing entries cleanly if needed
df.replace("", np.nan, inplace = True)
df.dropna(subset = ["arxiv_id"], inplace = True)

# Transform the DataFrame into a JSON string structure
# orient = 'records' ensure it builds an array of clean JSON objects
json_output = df.to_json(orient = "records", indent = 4)

# Print the first item as a quick validation preview
parsed_json = json.loads(json_output)
print(json.dumps(parsed_json[:1], indent = 4)) # Modify 1 to show more items
```
