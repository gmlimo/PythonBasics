# Arithmetic Operations on Images

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

img1 = plt.imread('/home/upbc/Pictures/misc/4.1.06.tiff')
img2 = plt.imread('/home/upbc/Pictures/misc/4.1.04.tiff')

plt.imshow(img1)
plt.show()
plt.imshow(img2)
plt.show()
```

```python
#Images must have same size (resolution)
plt.imshow(img1 + img2)
plt.show()
```

```python
plt.imshow(img1 - img2)
plt.show()
```

```python
#Invert values of the image matrix
plt.imshow(np.flip(img1, -1))
plt.show()
```

```python
plt.imshow(np.flip(img1, -2))
plt.show()
```

```python
plt.imshow(np.flip(img1, 0))
plt.show()
```

```python
plt.imshow(np.flip(img1, 1))
plt.show()
```

```python
plt.imshow(np.flip(img1, 2))
plt.show()
```

```python
#Roll (rotate right) values of the image matrix by 128 pixels
plt.imshow(np.roll(img1, 128))
plt.show()
```

```python
#Roll (rotate right) values of the image matrix by 256 pixels
plt.imshow(np.roll(img1, 256))
plt.show()
```

```python
#Flip image from left to right
plt.imshow(np.fliplr(img1))
plt.show()
```

```python
#Flip image from up-down
plt.imshow(np.flipud(img1))
plt.show()
```

```python
plt.imshow(np.rot90(img1))
plt.show()
```

# Logical Operations

```python
img1 = plt.imread('/home/upbc/Pictures/misc/4.1.06.tiff')
img2 = plt.imread('/home/upbc/Pictures/misc/4.1.04.tiff')

plt.imshow(np.bitwise_and(img1, img2))
plt.show()
```

```python
plt.imshow(np.bitwise_and(img2, img1))
plt.show()
```

```python
plt.imshow(np.bitwise_or(img1, img2))
plt.show()
```

```python
plt.imshow(np.bitwise_xor(img1, img2))
plt.show()
```

```python
plt.subplot(1, 2, 1)
plt.title('or')
plt.imshow(np.bitwise_or(img1, img2))
plt.subplot(1, 2, 2)
plt.title('xor')
plt.imshow(np.bitwise_xor(img1, img2))
plt.show()
```

```python
plt.imshow(np.bitwise_not(img1))
plt.show()
```

```python
plt.subplot(1, 2, 1)
plt.title('original')
plt.imshow(img2)
plt.subplot(1, 2, 2)
plt.title('not')
plt.imshow(np.bitwise_not(img2))
plt.show()
```

```python
img3 = plt.imread('/home/upbc/Pictures/misc/5.3.01.tiff')
#Grayscale image not operation
plt.subplot(1, 2, 1)
plt.title('original')
plt.imshow(img3, cmap = 'gray')
plt.subplot(1, 2, 2)
plt.title('not')
plt.imshow(np.bitwise_not(img3), cmap = 'gray')
plt.show()
```
