# Image Mask

```python
img1 = plt.imread('/home/upbc/Pictures/misc/4.1.06.tiff') 

#Some calculations
nrows, ncols, channels = img1.shape
row, col = np.ogrid[:nrows, :ncols] #Creates an open grid col = [0, 1, 2 ... ncols] and row = [0, 1, 2, ... nrows] (this last one in vertical array)
cnt_row, cnt_col = nrows/2, ncols/2 #Center of the image

#Image mask
outer_disk_mask = ((row - cnt_row) ** 2 + (col - cnt_col) ** 2 > (nrows/2) ** 2) 
img1 = np.copy(img1) #For modifying the values of an image matrix a copy must be created
img1[outer_disk_mask] = 0

plt.imshow(img1)
plt.axis('off')
plt.title('Masked Image')
plt.show()
```
