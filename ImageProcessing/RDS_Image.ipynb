# Read, Display and Save Images

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Read an image
img1 = plt.imread('/home/upbc/Pictures/misc/4.1.06.tiff')
```

```python
#Display an image
plt.imshow(img1)
plt.axis('off') #Axis won't show
plt.title('Tree')
plt.show()
```

```python
#Save image
plt.imsave('/home/upbc/Pictures/tree.png', img1)
```

```python
#A grayscale image, color is assigned
img2 = plt.imread('/home/upbc/Pictures/misc/5.3.01.tiff')
plt.imshow(img2)
plt.axis('off')
plt.show()
```

```python
#Display the image which is in grayscale format
img2 = plt.imread('/home/upbc/Pictures/misc/5.3.01.tiff')
plt.imshow(img2, cmap = 'gray')
plt.axis('off')
plt.show()
```
