```python
# Code that uses the Ollama library to run a local LLaMA model (e.g., LLaMA 3.2) for generating responses based on user prompts.
import json
import sys
import ollama

def run_with_model_and_template(model_filepath: str, template_filepath: str, new_prompt: str):
    # 1. Read the model name from the model configuration file
    try:
        with open(model_filepath, 'r', encoding='utf-8') as f:
            model_name = f.readline().strip()
            
        if not model_name:
            print(f"Error: The model file '{model_filepath}' is empty.", file=sys.stderr)
            return
            
        print(f"-> Selected model from file: {model_name}")

    except FileNotFoundError:
        print(f"Error: Model file '{model_filepath}' not found.", file=sys.stderr)
        return

    # 2. Load the training samples from the JSON template file
    try:
        with open(template_filepath, 'r', encoding='utf-8') as f:
            template_data = json.load(f)
            
        system_msg = template_data.get("system_instruction", "")
        samples = template_data.get("training_samples", [])
        print(f"-> Loaded {len(samples)} training samples from '{template_filepath}'.")

    except FileNotFoundError:
        print(f"Error: Template file '{template_filepath}' not found.", file=sys.stderr)
        return
    except json.JSONDecodeError as e:
        print(f"Error: Invalid JSON format in template file: {e}", file=sys.stderr)
        return

    # 3. Construct the conversation messages
    messages = []

    if system_msg:
        messages.append({"role": "system", "content": system_msg})

    for sample in samples:
        messages.append({"role": "user", "content": sample["input"]})
        messages.append({"role": "assistant", "content": sample["output"]})

    # Add the final prompt to execute
    messages.append({"role": "user", "content": new_prompt})

    # 4. Execute using Ollama
    try:
        print(f"-> Sending prompt and samples to {model_name}...")
        response = ollama.chat(
            model=model_name,
            messages=messages
        )
        
        print(f"\n=== {model_name} Response ===")
        print(response['message']['content'])
        print("==============================")

    except Exception as e:
        print(f"An error occurred communicating with Ollama: {e}", file=sys.stderr)

if __name__ == "__main__":
    model_file = "model_config.txt"
    template_file = "reviews.json"
    
    # Ensure the model file exists for this example
    with open(model_file, "w", encoding="utf-8") as f:
        f.write("llama3.2\n")

    # Ensure the template file exists for this example
    with open(template_file, "w", encoding="utf-8") as f:
        json.dump({
            "system_instruction": "You are a helpful assistant.",
            "training_samples": [
                {
                    "input": "What is the capital of France?",
                    "output": "The capital of France is Paris."
                }
            ]
        }, f)

    prompt_to_test = "Se aprecia un trabajo extenso que se valora, sin embargo se observan varias inconsistencias a lo largo del trabajo.  Se observa en la página 4 figura 1 la indicación de un modelo conceptual siendo este del tipo copo de nieve, sin embargo lo correcto sería indicar que es un esquema de diseño lógico y no de diseño conceptual de DW. Además un esquema conceptual correcto no debe indicar el tipo de dato. También en este aspecto se hace referencia a una bibliografía de esquema estrella. Faltó investigación sobre modelos conceptuales para DW para evitar lo anterior.  Mucha redundancia en problemas técnicos que no deberían ser de importancia para el artículo.  Se habla de inteligencia de negocios pero en un sentido estricto si se aplica inteligencia de negocios se necesita un  analista “experto” que sea capaz de tomar decisiones con las tendencias encontradas en el estudio, lo cual no se presenta en el artículo.  Faltan referencias de trabajos relacionados, pues solo se referencian  libros. Una revisión a trabajos relacionados habría sido muy útil en el sentido de estructurar mejor el artículo.  Existen metodologías para esquematizar el proceso ETL que habrían permitido mejorar este aspecto descrito en el artículo."
    
    run_with_model_and_template(model_file, template_file, prompt_to_test)
```
