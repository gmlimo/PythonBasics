# Double Ended Queue

```python
class Deque:
    def __init__(self):
        self.items = []
        
    def isEmpty(self):
        return self.items == []
    
    def size(self):
        return len(self.items)
    
    def addRear(self, item):
        self.items.insert(0, item)
        
    def addFront(self, item):
        self.items.append(item)
        
    def removeFront(self):
        return self.items.pop()
    
    def removeRear(self):
        return self.items.pop(0)
```

```python
d = Deque()
```

```python
d.isEmpty()
```

```python
d.addRear(2.71)
```
