```python
myTree = ['a',  #root
         ['b', #left subtree
           ['d', [], []],
           ['e', [], []] ],
         ['c', #right subtree
           ['f', [], []],
         [] ]
         ]
```

```python
print(myTree)
```

```python
print('Left Subtree = ', myTree[1])
print('Root = ', myTree[0])
print('Right Subtree = ', myTree[2])
```

```python
print('Root Element of the Right Subtree = ', myTree[2][0])
```
